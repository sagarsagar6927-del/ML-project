```python
import pandas as pd
from sklearn.datasets import load_iris
```

```python
iris = load_iris()
```

```python
irisIndData = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)
```

```python
irisIndData.head(5)
```

```python
import pandas as pd

irisIndData = pd.DataFrame(irisIndData)
```

```python

irisIndData.head(5)
from sklearn import tree
```

```python
dt = tree.DecisionTreeClassifier(max_depth=4, criterion='entropy')


from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    irisIndData,
    irisdata.target,
    test_size=0.2,
    random_state=0
)
```

```python
import graphviz
dt.fit(X_train, y_train)
dot_data = tree.export_graphviz(dt, out_file=None, feature_names=irisdata.feature_names, class_names=irisdata.target_names, filled=True, rounded=True, special_characters=True)
graph = graphviz.Source(dot_data)
graph
```
